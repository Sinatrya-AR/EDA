```python
# Data Loading
import pandas as pd

df = pd.read_csv("dataset_nilai_akademik_siswa.csv")

df.head()
```

```python
# Data Inspection
print("5 Data Pertama:")
display(df.head())

print("\nInformasi Dataset:")
df.info()

print("\nStatistik Dataset:")
display(df.describe())

print("\nUkuran Dataset:")
print(df.shape)

print(df.columns)
```

```python
# Cek Masalah Data
print("Jumlah Missing Value:")
print(df.isnull().sum())

print("Jumlah Data Duplikat:")
print(df.duplicated().sum())

print("Tipe Data:")
print(df.dtypes)
```

```python
# Data Cleaning

df = df.drop_duplicates()
print("Jumlah data setelah menghapus duplikat:", len(df))

df["nilai"] = (
    df["nilai"]
    .astype(str)
    .str.extract(r"(\d+\.?\d*)")[0]
)

df["nilai"] = pd.to_numeric(df["nilai"], errors="coerce")

df.loc[(df["nilai"] < 0) | (df["nilai"] > 100), "nilai"] = pd.NA
```

```python
# Menangani Missing Value nilai
df["nilai"] = df.groupby("mata_pelajaran")["nilai"].transform(
    lambda x: x.fillna(x.median())
)

df["nilai"] = df["nilai"].fillna(df["nilai"].median())
```

```python
# Menangani Missing Value guru_pengampu
df["guru_pengampu"] = df["guru_pengampu"].fillna("Tidak diketahui")
```

```python
# Menyeragamkan jenis_ujian
df["jenis_ujian"] = df["jenis_ujian"].str.upper()
```

```python
# Membersihkan tanggal_ujian
df["tanggal_ujian"] = pd.to_datetime(
    df["tanggal_ujian"],
    errors="coerce",
    format="mixed"
)

print(df["tanggal_ujian"].dtype)
```

```python
# Cek Ulang Setelah Cleaning
print("Missing Value Setelah Cleaning:")
print(df.isnull().sum())

print("\nDuplikat Setelah Cleaning:")
print(df.duplicated().sum())

print("\nTipe Data Setelah Cleaning:")
print(df.dtypes)

display(df.head())
```

```python
# Data Manipulation

# Filtering
data_uts = df[df["jenis_ujian"] == "UTS"]

display(data_uts)

# Sorting
data_sort = df.sort_values(by="nilai", ascending=False)

display(data_sort)
```

```python
# Kolom Turunan
def kategori_nilai(nilai):
    if nilai >= 90:
        return "Sangat Baik"
    elif nilai >= 80:
        return "Baik"
    elif nilai >= 70:
        return "Cukup"
    else:
        return "Perlu Perbaikan"

df["kategori_nilai"] = df["nilai"].apply(kategori_nilai)

display(df[["mata_pelajaran", "nilai", "kategori_nilai"]].head(10))
```

```python
# Groupby / Agregasi
rata_mapel = df.groupby("mata_pelajaran")["nilai"].agg(
    ["mean", "max", "min", "count"]
).round(2)

display(rata_mapel)
```

```python
# Groupby Berdasarkan Jenis Ujian
rata_ujian = df.groupby("jenis_ujian")["nilai"].mean().round(2)

display(rata_ujian)
```

```python
# Kalau Dataset Memiliki Kolom Kelas
rata_kelas = df.groupby("kelas")["nilai"].mean().round(2)

display(rata_kelas)
```

Berdasarkan hasil analisis awal, terdapat perbedaan rata-rata nilai antar mata pelajaran. Beberapa mata pelajaran memiliki rata-rata nilai yang lebih tinggi dibandingkan mata pelajaran lainnya. Selain itu, rata-rata nilai juga berbeda berdasarkan jenis ujian. Data yang telah dibersihkan dapat digunakan untuk melihat pola nilai siswa berdasarkan mata pelajaran dan jenis ujian.
